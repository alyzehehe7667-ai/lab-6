```python

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
df.head()
```

```python

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

print("Missing values per column:")
print(missing)

print("\nMissing percentage:")
print((missing / len(df) * 100).round(1))
```

```python

numeric = df.select_dtypes(include="number")

cv = (numeric.std() / numeric.mean()).abs().sort_values(
    ascending=False
)

print("Coefficient of variation:")
print(cv.round(3))

print("\nLargest:", cv.index[0])
```

```python

def inspect(path_or_url):
    import pandas as pd

    data = pd.read_csv(path_or_url)

    print("=" * 60)
    print("SOURCE:", path_or_url)
    print("SHAPE:", data.shape[0], "rows x", data.shape[1], "columns")

    print("\n--- DATA TYPES ---")
    print(data.dtypes)

    print("\n--- MISSING VALUES ---")
    missing = data.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("None")
    else:
        report = pd.DataFrame({
            "missing": missing,
            "percent": (missing / len(data) * 100).round(1)
        })
        print(report.sort_values("missing", ascending=False))

    print("\n--- NUMERIC SUMMARY ---")
    print(data.describe())

    return data


# Dataset 1: Titanic
titanic = inspect(
    "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
)
```

```python

iris_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

iris = inspect(iris_url)
```
