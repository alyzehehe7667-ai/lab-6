```python

# STEP 1: Extract project ZIP
!unzip -o /content/Diabetes_Prediction_Lab_GitHub.zip -d /content/diabetes_lab_project

# STEP 2: Go to project folder
%cd /content/diabetes_lab_project

# STEP 3: Install required libraries
!pip install -r requirements.txt

# STEP 4: Import libraries
import os
import pickle
import numpy as np

# STEP 5: Check project files
print("\n--- PROJECT FILES ---")
print(os.listdir("."))

# STEP 6: Load trained model
with open("Mymdoel.pkl", "rb") as f:
    model = pickle.load(f)

print("\n--- MODEL INFORMATION ---")
print("Model loaded successfully!")
print("Model type:", type(model))
print("Required input features:", model.n_features_in_)
print("Possible output classes:", model.classes_)

# STEP 7: Test prediction with placeholder inputs
sample_input = np.zeros((1, model.n_features_in_))
prediction = model.predict(sample_input)

print("\n--- TEST PREDICTION ---")
print("Prediction output:", prediction)

if hasattr(model, "predict_proba"):
    print("Prediction probabilities:", model.predict_proba(sample_input))
```
